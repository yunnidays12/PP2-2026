## 상속 : "is-a 관계"
* 기존에 존재하는 클래스로부터 코드와 데이터를 이어받고 자신이 필요한 기능을 추가하는 기법.
* 부모 클래스와 자식 클래스가 있으며, 부모 클래스가 자식 클래스를 포함하고 있는 구조를 가지고 있다.
* 그래서 부모 클래스를 잘 구성해야, 자식 클래스를 쉽게 만들 수 있다.
---

```python
class Car : #부모클래스
    def __init__ (self, maker, model, color, price) :
        self.maker = maker
        self.model = model
        self.color = color
        self.price = price

    def setMaker(self, maker) :
        self.maker = maker

    def getMaker(self) :
        return self.maker

    def get_info(self) :
        return self.__str()

    def __str__(self) :
        return f"Maker : {self.maker}, Model : {self.model}, Color : {self.color}, Price : {self.price}"

class ElectricCar(Car) : #클래스(부모클래스)
    
    def __init__ (self, maker, model, color, price, batterySize) :
        super().__init__(maker, model, color, price) #이미 부모클래스에서 maker, model, color, price는 다 정의되있다.
        self.batterySize = batterySize
        
    #아무것도 안 쳐도 setMaker, getMaker도 다 정의되있다.
    def setBattery(self, battery_size) :
        self.batterySize = battery_size

    def getBattery(self) :
        return self.batterySize

    def __str__ (self) :
        return f"{super().__str__()}, BatterySize : {self.batterySize}"

    def get_info(self):
        return self.__str__()

class Truck(Car) :
    def __init__ (self, maker, model, color, price, max_carrying) :
        super().__init__(maker, model, color, price)
        self.max_carrying = max_carrying

    def __str__ (self) :
        return f"{super().__str__()}, Max carrying : {self.max_carrying}"


def main() :
    mycar = ElectricCar("Tilsa", "Model S", "white", 10000, 0)
    mycar.setMaker("Tesla")
    mycar.setBattery(40)
    print(mycar.get_info())

    mytruck = Truck("기아", "봉고3", "흰색", "13848294", "1톤")
    mytruck.setMaker("KIA")
    print(mytruck)

    mytruck.getMaker()
    print(mytruck.getMaker())

if __name__ == "__main__" :
    main()
```

## 다중상속 : 2개 이상의 부모클래스로부터 상속받을 때

```python
class Base1 :
    def __init__(self, v1=10):
        self.attr1= v1

    def display1(self) :
        print(f"{self.attr1}")

class Base2 :
    def __init__(self, v2=20):
        self.attr2= v2
    def display2(self) :
        print(f"{self.attr2}")

class C(Base1, Base2) :
    def __init__ (self, v1, v2, v3) :
        Base1.__init__(self, v1)
        Base2.__init__(self, v2)
        self.attr3 = v3

    def __str__(self):
        return f"{(self.attr1, self.attr2, self.attr3)}"

    def print(self) :
        self.display1()
        self.display2()

def test() :
    c = C(v1=1, v2=3, v3=4)
    c.print()

if __name__ == "__main__" :
    test()
```
