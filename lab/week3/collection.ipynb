## Tuple 
> 수정 불가능한 리스트. 불변하는 성질을 지닌다. (추가는 가능)

* 튜플을 만들 때는 "mytuple = ()"을 이용해 정의한다.
* 주민등록번호와 같이 극도로 중요한 정보들을 모을 때 쓴다.
* 튜플 <--> 리스트 : 서로 변환이 가능하다.
    * 튜플 --> 리스트 : list(mytuple)
    * 리스트 --> 튜플 : tuple(mylist)
* 변경은 안되지만, "추가"는 가능하다.
    * mytuple1 = mytuple1 + mytuple2
* 튜플의 데이터를 패킹할수도, 언패킹할 수도 있다.
    * 패킹
    * 언패킹

## set
> 수학에서의 집합과 같은 개념 (순서x, 중복x)

* 세트를 만들 때는 "myset = set()"을 이용해 정의한다. ("()"을 쓰면 tuple로 인식함에 주의)
* 중복과 순서를 고려하지 않을 때 사용하는 자료구조이다.
* set의 메소드는 다음과 같다.
    * myset.discard(data) : data를 집합에서 제거한다.
    * myset.add(data) : data를 집합에 추가한다.
        * 중복되면 자동으로 걸러주는 기능이 있다.
    * myset1 & myset2 : 두 집합의 "교집합"을 출력한다.
    * myset1 | myset2 : 두 집합의 "합집합"을 출력한다.
    * myset1-myset2 : 두 집합의 "차집합"을 출력한다.

## dictionary
> key값과 value값을 1:1로 매핑. (순서x, "키"의 중복x)

* 딕셔너리를 만들때는 "my_dict = {}"을 이용해 정의한다.
* 이름:전화번호, 이름 : 주민등록번호와 같이 1:1로 매핑되게 저장된다.
* 데이터를 찾을 때, 리스트와 다르게 O(N)의 시간복잡도를 갖는다.
* 해쉬 테이블을 이용해서 데이터를 빠르게 찾을 수 있다.

```python
#327p. 학생 성적 처리
score_dic = {
    "kim" : [99, 83, 95],
    "Lee" : [68, 45, 78],
    "Choi" : [25, 56, 69]
}

for name in score_dic :
    sum_of = 0
    for score in score_dic[name] :
        sum_of += score
    print(f"{name}의 평균성적 : {sum_of/len(score_dic[name]):.3f}")
```

```python
#335p. 단어 카운터 만들기
text_data = "Create the highest, grandest vision possible for your life, becase you become what you believe"

db = {}
list_word = list(text_data.split())

for word in list_word :
    if word in db : db[word] += 1
    else : db[word] = 1

for word in db :
    print(f"{word} 의 등장횟수 = ", db[word])
```

```python
#343p. 회문 검사하기
user_input = input("문자열을 입력하세요. : ")

#[방식1] 엄청 복잡하게 구현하기
reverse = ""
for i in range(len(user_input), -1, -1) :
    reverse += user_input[i]
if reverse == user_input : print("회문입니다.")
else : print("회문이 아닙니다.")

#[방식2] 꽤 간편하게 구현하기
if user_input == user_input[::-1] : print("회문입니다.")
else : print("회문이 아닙니다.")
```

```python
#350p. 문자열 분석
db = {"digits" : 0, "spaces" : 0, "alphas" : 0}
user_input = input("문자열을 입력하시오. : ")
for i in user_input :
    if i.isdigit() == True : db["digits"] += 1
    elif i == " " : db["spaces"] += 1
    else : db["alphas"] += 1

print(db)
```
