### 리스트
> 가변변수 중 한 자료구조
* 파이썬의 리스트는 C언어의 배열(array)와 다르게, 객체지향 자료구조이다.
* 이 리스트는 "순서"대로 자료가 들어가있다. 이때의 자료형은 어떤 자료형도 가능하다.
* 리스트는 집합(set)와는 다르게, 리스트 내 같은 원소가 있더라도 인덱스 번호가 다르면 다른 자료이다.
* 리스트는 "0번"부터 번호를 붙여서 사용한다. (1번부터 아님에 주의.)
* 뒤에서부터 읽고 싶다면 name_list[-i]와 같이 사용한다. (0번부터 아님에 주의.)
* "list index out of range"가 뜨지 않도록 유의해야한다. (번호 잘 붙이기.)
---
### 리스트 방문
* 인덱스를 이용하는 방법
    * 의미 : 리스트의 "i번째" 원소에 접근한다!
    * 예 : for i in range(0, len(name_list))
* for-each 루프를 이용하는 방법
    * 의미 : 리스트 안에 있는 원소에 접근한다!
    * 예 : for element in name_list
* 리스트에 자료가 있는가를 확인하고 싶다면? : in
    * 예 : if (검사하고 싶은 값) in name_list (있다.)
    * 예 : if (검사하고 싶은 값) not in name_list (없다.)
---
### 리스트 메소드
* list_name.append(data) : data를 list_name에 추가한다. (오른쪽 끝에)
* list_name.insert(index_num, new_data) : index_num에 new_data를 추가한다. (중간에 끼워넣기)
* list_name.index(data) : list_name에 data의 번호를 반환한다. (없으면 오류.)
* list_name.pop(index) : list_name에 index번째 데이터를 삭제한다.
* list_name.remove(data) : list_name에 data를 찾아서 해당 데이터를 삭제한다.
    * 질문 : "cherry"가 리스트에 두개였는데, .remove("cherry")를 하면 둘 다 삭제? or 하나만 삭제?
* random.choice(list_name) : random 모듈을 이용해서, list_name에 있는 원소 중 랜덤으로 하나 택하는 모듈
---
### 리스트 함수
* len(list_name) : list_name의 길이를 반환한다.
* sum(list_name) : list_name의 합을 반환한다.
* min(list_name), max(list_name) : list_name의 최대, 최소를 반환한다.
* sorted(list_name) : : list_name의 정렬된 리스트를 반환한다.
---
### 리스트 합병과 복제 그리고 비교
* list1 + list2 : [data1, data2] + [data3, data4] = [data1, data2, data3, data4]
* [data1, data2, ... , data_n] * m = [data1, data2, ... , data_n, data1, data2, ... , data_n, ...]
    * 똑같은 순서대로 n개가 반복되서 출력된다.
* [data] * n = [data, data, data, ..., data] (총 n개가 리스트에 담김)
* list1 == list2 : list1의 모든 원소와 list2의 모든 원소가 '순서'까지 똑같아야 함.
* list1 > list2 : list1의 모든 "정수형" 자료의 원소가 list2의 같은 인덱스의 모든 "정수형" 자료의 원소보다 크다.
* list2 = list1 : list1의 주소만 list2가 복사하는 명령어. list1이 바뀌면 list2도 바뀜.
* list2 = list(list1) : list1 자체를 복제하는 명령어. list1이 바뀌어도 list2는 바뀌지 않음.

#### 리스트 안에 리스트의 복제
* 만약, list1 = [[1,2], 3, 4, 5]였고, list2 = list(list1)이라고 하면, list2는 list1와 독립일까?
    * 결과 : 그렇지 않다. (아래의 코드에서 확인)
* 2차원 이상의 리스트에서는 참조(주소)만 복사한다.

```python
list1 = [[1, 2], 3, 4, 5]
list2 = list(list1)

print(list1)
print(list2)

list1[0][1] = 5

print(list1)
print(list2) #얕은 복사의 한계
```

### 2차원 리스트

```python
#265p. 성적 처리 프로그램
score = [] #[10,20,60,70,80] 입력
stu_cnt = int(input("학생의 수를 입력하세요. : "))
upper_80 = 0

for _ in range(stu_cnt) :
    stu_score = float(input("학생의 성적을 입력하세요. : "))
    score.append(stu_score)
    if stu_score >= 80 : upper_80 += 1

print("성적평균 : ", sum(score)/stu_cnt)
print("최대점수 : ", max(score))
print("최소점수 : ", min(score))
print("80점 이상 : ", upper_80)
```

```python
#269p. 친구관리 프로그램
def print_info() :
    print("----------------------")
    print("1. 친구 리스트 출력")
    print("2. 친구추가")
    print("3. 친구삭제")
    print("4. 이름변경")
    print("9. 종료")
    print("----------------------")

def print_list(friend_list) :
    print(friend_list)

def add_friend() :
    friend_name = input("이름을 입력하세요. : ")
    return friend_name

def remove_friend(friend_list) :
    friend_name = input("삭제할 친구의 이름을 입력하세요.")
    if friend_name in friend_list :
        fruit_list.remove(friend_name)
        print("친구삭제가 되었습니다.")
        
    else :
        print("입력한 이름이 없습니다.")
    
    return friend_list

def change_name(friend_list) :
    friend_name = input("원래 친구의 이름은? : ")
    if friend_name in friend_list :
        new_name = input("바꿀 친구의 이름은? : ")
        friend_list.remove(friend_name)
        friend_list.append(new_name)
    else : print("친구 리스트에 없습니다.")
    return friend_list

def main() :
    friend_list = []

    while True :
        print_info()
        user_choice = int(input("메뉴를 선택하세요. : "))

        if user_choice == 1 :
            print_list(friend_list)
        elif user_choice == 2 :
            friend = add_friend()
            friend_list.append(friend)
        elif user_choice == 3 :
            friend_list = remove_friend(friend_list)
        elif user_choice == 4 :
            friend_list = change_name(friend_list)
        elif user_choice == 9 :
            print("프로그램을 종료합니다.")
            break
    
if __name__ == "__main__" :
    main()
```
