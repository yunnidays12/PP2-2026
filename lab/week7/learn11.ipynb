```python
#LAB : 내장함수 예제

"""
[문제]
  - 우리는 파티에 적합한 양의 음식을 구입하기 위해서, 총 인원이 몇 명인지 알아야한다.
  - 파티에 초대를 한 사람들의 리스트 invitations와 각 초대자가 초대한 인원이 담긴 정수 리스트 persons가 주어진다.
  - (1) 파티에 한 사람이라도 오는 지를 확인해보자.
  - (2) 모든 초대받은 그룹이 전부 오는지 확인해보자.
  - (3) 가장 많이 오는 그룹에는 몇 사람이나 있는지 알아보고, 초대자의 이름을 출력하시오.
"""

def solution(invitations, persons) :
    print(any(persons)) # (1) 한 사람이라도 오면 True
    print(all(persons)) # (2) 한 그룹이라도 0명이라면 False

    # (3) 람다함수로 정렬하기.
    new_list = list(zip(invitations, persons))
    answer_list = sorted(new_list, key=lambda x : x[1])
    print(answer_list[-1])
    
    
invitations = ["Kim", "Lee", "Park", "Choi"]
persons = [1, 3, 0, 6]
solution(invitations, persons)
```

```python
#LAB : 키를 이용한 정렬 예제
"""
[문제]
  - 주소록을 작성한다고 하자. 간단하게 사람들의 이름과 나이만 저장할 것이다.
  - 사람을 Person이라는 클래스로 나타낸다고 할 때, 인스턴스 변수는 다음과 같다.
    * name : 이름 (문자열) / age : 나이 (정수형)
  - 나이순으로 정렬하여 보여주는 프로그램을 작성하자.
"""

class Person :
    def __init__ (self, name, age) :
        self.name = name
        self.age = age

    def __repr__(self):
        return repr((self.name, self.age))

def solution() :
    persons = [Person("홍길동", 20), Person("김철수", 35), Person("최자영", 38)]
    answer_list = sorted(persons, key=lambda x : x.age)
    for person in answer_list :
        print(f"<이름 : {person.name}, 나이 : {person.age}>", end=",")

solution()
```

```python
#LAB : 람다식으로 온도 변환하기
"""
[문제]
  - 람다식과 map() 함수를 이용해서 화씨온도를 섭씨온도로 변환하는 명령문을 작성하라.
  - 공식은 다음과 같다. : (°F − 32) × 5/9
"""

def solution(c_temp) :
    return list(map(lambda x : (5.0/9.0)*(x-32.0), c_temp))

c_temp = [0,10,20,30,40,50]
solution(c_temp)
```

```python
#LAB: 람다식으로 데이터 처리하기
"""
[문제]
  - 순서대로 (주문번호, 물품, 수량, 품목 당 가격)을 원소로 갖는 튜플 리스트 store_data가 주어질 때,
  - 람다식을 이용하여 (주문번호, 수량*품목 당 가격)을 반환하는 프로그램을 작성하라.
"""

def solution(store_data) :
    answer = list(map(lambda x : (x[0], x[2]*x[3]) , store_data))
    print(answer)

store_data = [(1, "재킷", 5, 120000), (2, "셔츠", 6, 24000)]
solution(store_data)
```

```python
#LAB : 피보나치 이터레이터
"""
[문제]
  - 피보나치 수열이란 앞의 두 수의 합이 바로 뒤의 수가 되는 수열을 의미한다.
  - 피보나치 수열의 수들을 생성하는 이터레이터 클래스를 정의해보자.
"""

class Fibo :
    def __init__(self, a=1, b=0, maxValue = 50):
        self.a = a
        self.b = b
        self.maxValue = maxValue

    def __iter__(self) :
        return self

    def __next__ (self) :
        n = self.a + self.b
        if n > self.maxValue :
            raise StopIteration()
        self.a = self.b
        self.b = n
        return n

for i in Fibo() :
    print(i, end=" ")
```

```python
#LAB : Book 클래스
"""
[문제]
  - 책을 나타내는 Book 클래스를 작성하라.
  - Book 클래스 내부에 __add__() 함수를 정의해서, Book 클래스의 객체를 서로 합칠 수 있게 하라.
  - __add__() 함수는 책의 페이지들을 합쳐서 반환한다.
"""

class Book :
    def __init__(self, name, page):
        self.name = name
        self.page = page

    def __add__(self, other):
        return self.page + other.page

book1 = Book('책1', 600)
book2 = Book('책2', 700)

print(book1+book2)
```

```python
#LAB: 동전 던기지 게임

"""
[문제]
  - 동전 던지기 게임을 파이썬으로 작성해보라. random 모듈을 사용한다.
"""

import random

def solution() :

    while True :

        user_input = input("동전 던지기를 계속하시겠습니까?(yes, no) : ")

        if user_input == "yes" :
            result = ["head", "tail"]
            print(random.choice(result))

        else : break

solution()
```
