```python
#Counter 클래스 만들기

"""
    - 데이터 : 현재 숫자
    - 동작
        ⓐ 사용자가 입력하는 숫자만큼 증가하기
        ⓑ 현재의 값을 읽어오기
"""

class Counter : #클래스의 이름을 정의할 때는 항상 대문자로 시작함이 원칙이다.
    def __init__ (self) : #생성함수
        self.count = 0

    def increment(self, value = 1) : #초기값을 주고 싶다면? : defult 변수를 사용하기.
        #만약에 value가 주어지지 않았다면 그냥 1이 value가 될 것이다.
        self.count += value

    def get_count(self) :
        return self.count

def test_counter() :
    counter1 = Counter()
    counter2 = Counter()

    n1 = int(input("1번 카운터에 증가시킬 수를 입력하세요. : ")) #50입력
    n2 = int(input("2번 카운터에 증가시킬 수를 입력하세요. : ")) #60입력

    counter1.increment(n1)
    counter2.increment(n2)

    print(counter1.get_count())
    print(counter2.get_count())
    
if __name__ == "__main__" :
    test_counter()
```

```python
#p.370 TV 클래스 만들기
class MyTelevision :

    """
        - 데이터 : 채널번호, 볼륨, 전원상태
        - 동작
            * 전원을 키고, 끄는 행위
            * 채널을 바꾸는 행위
            * 볼륨을 바꾸는 행위
            * 현재 티비의 상태를 출력하는 행위
    """

    serialNumber = 0
    def __init__(self, number = 1, vol = 10, is_on = False):
        self.number = number
        self.vol = vol
        self.is_on = is_on
        MyTelevision.serialNumber += 1 #자동으로 serialNumber 가 올라가서, 자동으로 시리얼 번호가 부여됨.
        self.myserial = MyTelevision.serialNumber

    def turnOn(self) :
        self.is_on = True

    def turnOff(self) :
        self.is_on = False

    def changeChannel(self, number=11) :
        self.number = number

    def changeVolume(self, vol=5) :
        self.vol = vol

    def current_state (self) :
        print("티비의 전원 : ", self.is_on)
        print("티비의 채널번호 : ", self.number)
        print("티비의 볼륨 : ", self.vol)

class Remote :

    """
    - 데이터 : 없음
    - 동작
        * 연관된 TV를 켜거나 끄라는 동작만 지원
    """
    
    def __init__ (self, object : MyTelevision) :
        self.object = object

    def object_change(self, object) :
        self.object = object

    def pressOn(self) :
        self.object.turnOn()

    def pressOff(self) :
        self.object.turnOff()

    def press_display(self) :
        self.object.current_state()

def tv_test() :
    tv1 = MyTelevision(number=11)
    remote = Remote(object=tv1)

    remote.pressOn()
    remote.press_display()
    remote.pressOff()
    remote.press_display()

if __name__ == "__main__" :
    tv_test()
```

```python
class Circle :
    def __init__(self, radius):
        self.radius = radius

    def get_radius(self) :
        return self.radius

    def __eq__(self, other) :
        return self.radius == other.radius

c1 = Circle(4)
c2 = Circle(5)

print(c1 == c2)
print(c1<c2) #[출력결과] TypeError: '<' not supported between instances of 'Circle' and 'Circle'
```

```python
#LAB : 벡터의 연산
class Vector :
    def __init__(self, x=0, y=0) :
        self.x = x
        self.y = y
    def __add__ (self, other) :
        return (self.x+other.x, self.y+other.y)
    def __sub__ (self, other) :
            return (self.x-other.x, self.y-other.y)
    def __eq__ (self, other) :
            return (self.x == other.x) and (self.y == other.y)
```

```python
#LAB : 은행계좌

class Account :
    def __init__(self, money = 1000) :
        self.__balance = money

    def withdraw(self, m1) :
        self.__balance -= m1
        print(f"통장에서 {m1}이 출금되었음.")

    def deposit(self, m2) :
        self.__balance += m2
        print(f"통장에 {m2}이 입금되었음.")
```

### Person 객채 제작하기 (실습)
 * 데이터 : 이름, 출생연도
 * 행동
   * 이름을 출력하게 하기
   * 올해 나이를 출력하게 하기
 * 테스트 함수
   * 우리 팀의 4명에 대한 객체를 생성하고 목록에 저장

```python
class Person :
    def __init__(self, name, year):
        self.__name = name
        self.year = year

    def get_name(self) :
        return self.__name

    def print_info(self) :
        print(f"{self.__name}님은 {2026 - self.year}살입니다.")

    def __eq__ (self, other) :
        if self.year == other.year : return "동갑"
        else : "동갑이 아님"

def test_function() :
    team = []

    n = int(input("팀원이 몇 명인지 입력하세요. : "))

    for _ in range(n) :
        name, age = map(str, input("팀원의 이름과 태어난 년도를 공백을 기준으로 입력하세요. : ").split())
        team.append([name, int(age)])

    for person in team :
        people = Person(name=person[0], year=person[1])
        people.print_info()

if __name__ == "__main__" :
    test_function()
```
