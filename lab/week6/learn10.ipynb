## 파일 입출력 연습 노트 파일입니다.

이 파일은 파이썬의 파일 입출력 예제와 이론을 다루는 노트입니다.
* 개념과 문제의 수준
    * 개념 : 수업시간에 배운 것을 기록 후, 공부하며 깨닫는 것들 remark
    * 문제 : 개념에 맞는 문제를 직접 제작하거나, 책에 있는 연습문제 풀이.

```python
#LAB : 매출 파일 관리

"""
[요구조건]
  1. 입력 파일 (sales.txt)에 일별 매출이 저장되어 있다.
  2. 이것을 읽어서 일별 평균 매출과 총 매출을 계산하고,
  3. 그 매출을 다른 파일에 출력하는 프로그램을 작성하라.

[입력]
  1. 하루 매출이 한 줄에 정수로 기록되어 있다.

[출력]
  1. 총매출과 평균 일매출이 한 줄에 각각 출력된다.
"""

def read_file(path) :

    total = 0
    cnt = 0

    with open(path, "r", encoding="utf-8") as f :
        data = f.readlines()
        for line in data :
            clean_line = line.strip()
            total += float(clean_line)
            cnt += 1
    
    return total, cnt

def write_func(total, cnt) :

    with open("lab\week6\매출파일처리\summary.txt", "w", encoding="utf-8") as f :
        f.write(f"총매출 : {total}\n")
        f.write(f"평균 일매출 : {total/cnt}")

if __name__ == "__main__" :
    path = "lab\week6\매출파일처리\sales.txt" #일단은 상대경로로 했는데 말을 안들어서 패스
    total, cnt = read_file(path)
    write_func(total, cnt)
```

```python
#LAB : 행맨

"""
[요구조건]
  1. 입력 파일 (hangman.txt)에 여러 개의 단어가 적혀있다.
  2. 이것을 읽어서 랜덤으로 하나의 단어를 택하고,
  3. 유저가 한 글자씩 스펠링을 입력했을 때, 맞으면 글자가 보이고, 아니면 시도횟수만 하나 증가하게 만들어라.

[입력]
  1. 한 줄에 한 단어씩 줄바꿈 기호를 기준으로 구분되어 주어진다.
[출력]
  1. 기회는 10번이고, 맞췄다면 글자가 한 글자 보이고, 틀렸다면 시도횟수만 하나 깎이게 한다.
  2. 유저가 2글자 이상을 입력했다면 1글자씩만 입력하라는 예외처리를 한다.
  3. 유저가 공백을 입력했다고 해도 글자를 입력하라는 예외처리를 한다.

"""

import random

def select_word(path) :
    word_list = []
    with open(path, "r", encoding="utf-8") as f :
        data = f.readlines()

        for word in data :
            clean_word = word.strip().split("\n")
            word_list.append(clean_word)

    return random.choice(word_list)[0]

def hangman(word) :
    cnt = 0
    chance = 10
    answer = ""

    print(f"단어의 스펠링의 개수는 {len(word)}개입니다.")

    while True :
        print("=========")
        user_input = input(f"단어를 추측하세요. : {answer}")
        if len(user_input) == 1 and user_input != " " :
            if user_input == word[cnt] :
                cnt += 1
                answer += user_input
                print("맞았습니다!")

            else :
                print("틀렸습니다.")
                chance -= 1
                print(f"기회는 {chance}번 남았습니다.")

            if chance == 0 :
                print("기회가 모두 소진되었습니다.")
                print(f"정답은 {word}이었습니다.")
                break

            if cnt == len(word) :
                print(f"정답입니다! 정답은 {word}이었습니다.")
                break
        else : 
            print("한 글자씩만 입력하세요.")

def main() :
    path = "lab\\week6\\행맨\\hangman.txt"
    word = select_word(path)
    hangman(word)

if __name__ == "__main__" :
    main()
```

```python
#LAB : 각 문자 횟수세기

"""
[요구조건]
  1. 입력 파일 (string_count.txt)에 여러 줄의 문장이 적혀있다.
  2. 이것을 읽어서 각 문자들이 몇 번이나 나타나는지를 세는 프로그램을 작성하라.
  3. 여기서는 딕셔너리를 이용하라.

[입력]
  1. 여러 줄에 걸쳐 (혹은 한 줄로) 문장이 하나 주어진다. 

[출력]
  1. 각 문자들이 몇 번이나 나타나는지를 딕셔너리 원형의 상태로 출력한다.
"""

def read_file(path) :

    db = {}
    
    with open (path, "r", encoding="utf-8") as f :
        data = f.readlines()

        for line in data :
            clean_line = line.strip()
            for c in clean_line :
                if c in db : db[c] += 1
                else : db[c] = 1

    return db

def main() :
    path = "lab\week6\string_count.txt"
    print(read_file(path))

if __name__ == "__main__" :
    main()
```
