```python
#레이블이 있는 윈도우를 생성해보자.

"""
[참고] window : Tk 클래스의 객체가 형성된다.

[참고] geometry 
  1. 단위는 pixel이고, 크기는 "문자열"로 입력한다.
  2. 곱하기는 소문자 x로 대체.
  3. (0,0)은 좌측상단을 기준으로 한다. 수학처럼 사고 금지.

[참고] label : window가 label이라는 클래스의 부모 클래스인 상태.

[참고] 컨테이너 : 위젯이나 컴포넌트를 올려놓고 담아둘 물리적인 판의 역할을 함.

[참고] mainloop()의 역할
  1. 사용자로부터 오는 마우스나 키보드 이벤트를 "순서대로" 처리한다.
  2. 이런 자료구조를 "큐(queue)"라고 한다. 선입선출의 형태를 띈다.

[주의] 쓸 때 주의할 것.
  1. mainloop() 후의 코드들은 mainloop()가 종료되기 전까지 실행X
  2. 루프 안에서 무거운 작업 금지.
"""

from tkinter import *

window = Tk() #Tk 클래스의 객체가 형성. 
window.title ("Hello, 김윤성.")
window.geometry("300x200") # [단위] pixel / [주의] 곱하기는 소문자 x로 대체.
#[참고] (0,0)은 좌측상단을 기준으로 한다.

label = Label(window, text = "Hello tkinker") #라벨 위젯을 생성한다.
label.pack() #라벨을 컨테이너에 위치시킨다.

window.mainloop()
```

```python

#[요구조건] 버튼을 누르면 콜백 신호가 돌아오도록 한다.

from tkinter import *

window = Tk()
window.title("OOButton")
window.geometry("600x400")

label = Label(window, text = "Break your limit").pack()

# [버튼] 함수와 command를 이용해서 신호가 돌아오도록 하기.

cnt = 0

def on_btn1_cilcked() :
    global cnt
    # [출력결과 中] 버튼이 클릭되었습니다. : <ButtonPress event state=Mod1 num=1 x=22 y=17>
    cnt += 1
    label.config(text = f"버튼1이 클릭되었습니다. : {cnt}번")
    # [출력결과] label에 버튼1이 클릭되었다와 cnt 을 출력
    
# btn1 = Button(window, text = "Click your limit.", bg = "#000000", fg = "#ffffff", command = on_btn1_cilcked)
btn1 = Button(window, text = "Click your limit.", bg = "#000000", 
              fg = "#ffffff", width = 30, height=20)

btn1.bind("<Button-1>", on_btn1_cilcked)
btn1.pack()

window.mainloop()
```

```python
from tkinter import *
window = Tk()

Label(window, text = "섭씨").grid(row=1, column=0)
Label(window, text = "화씨").grid(row=0, column=0)

def compute() :
    degree = float(e1.get().strip())
    tc = (degree-32.0)*5.0/9.0
    e2.delete(0, END)
    e2.insert(0, str(tc))

e1 = Entry(window)
e1.grid(row=0, column=1)
e2 = Entry(window)
e2.grid(row=1, column=1)

b1 = Button(window, text = "화씨 -> 섭씨", command=compute)
b1.grid(row=2,column=1)

window.mainloop()
```

```python
import random
from tkinter import *

window = Tk()
answer = random.randrange(1, 101)

count = 0

def tryed():
    global answer, count
    count += 1  # 몇 번 눌렸는지 카운트
    
    answered = int(user_input.get())
    
    if answered > answer:
        msg = "높음!"
    elif answered < answer:
        msg = "낮음!"
    else:
        msg = "정답!"

    # 텍스트 뒤에 시도 횟수를 붙여서 화면이 갱신되는지 확인
    result['text'] = f"{msg} (시도: {count}회)"
    user_input.delete(0, END)

def reset() :
    global answer, count
    answer = random.randrange(1, 101)
    count = 0
    result['text'] = "다시 리셋했어요."

label = Label(window, text = "숫자 게임에 오신 것을 환영합니다!").grid(row=0, column=0)
user_input = Entry(window)
user_input.grid(row=1, column=0)

try_button = Button(text="시도", fg = "green", command=tryed)
try_button.grid(row=1, column=1)

reset_button = Button(text="초기화", fg="red", command=reset)
reset_button.grid(row=1, column=2)

result = Label(text="1부터 100 사이의 숫자를 입력하시오.")
result.grid(row=1, column=3)

window.mainloop()
```
