## GUI 프로그래밍 연습 노트 파일입니다.

이 파일은 파이썬의 GUI 프로그래밍 예제와 이론을 다루는 노트입니다.
* 개념과 문제의 수준
    * 개념 : 수업시간에 배운 것을 기록 후, 공부하며 깨닫는 것들 remark
    * 문제 : 개념에 맞는 문제를 직접 제작하거나, 책에 있는 연습문제 풀이.

* 작성일자 : 2026.10.27

```python
#레이블이 있는 윈도우를 생성해보자.

"""
[참고] window : Tk 클래스의 객체가 형성된다.

[참고] geometry 
  1. 단위는 pixel이고, 크기는 "문자열"로 입력한다.
  2. 곱하기는 소문자 x로 대체.
  3. (0,0)은 좌측상단을 기준으로 한다. 수학처럼 사고 금지.

[참고] label : window가 label이라는 클래스의 부모 클래스인 상태.

[참고] 컨테이너 : 위젯이나 컴포넌트를 올려놓고 담아둘 물리적인 판의 역할을 함.

[참고] mainloop()의 역할
  1. 사용자로부터 오는 마우스나 키보드 이벤트를 "순서대로" 처리한다.
  2. 이런 자료구조를 "큐(queue)"라고 한다. 선입선출의 형태를 띈다.

[주의] 쓸 때 주의할 것.
  1. mainloop() 후의 코드들은 mainloop()가 종료되기 전까지 실행X
  2. 루프 안에서 무거운 작업 금지.
"""

from tkinter import *

window = Tk() #Tk 클래스의 객체가 형성. 
window.title ("Hello, 김윤성.")
window.geometry("300x200") # [단위] pixel / [주의] 곱하기는 소문자 x로 대체.
#[참고] (0,0)은 좌측상단을 기준으로 한다.

label = Label(window, text = "Hello tkinker") #라벨 위젯을 생성한다.
label.pack() #라벨을 컨테이너에 위치시킨다.

window.mainloop()
```

```python
# 버튼 위젯이 있는 윈도우를 생성해보자.
# 색을 정할 때는 "#000000"와 같은 HEX code를 이용하면 편함.

from tkinter import *
window = Tk()

button = Button(window, 
                
    text = "클릭하세요.", #버튼에 들어갈 글자.
    bg = "#AC30FF", #버튼의 배경색깔
    fg = "#FFFFEF", #글자의 색
    width = 80, height= 2 #버튼의 크기 설정 (글자수 기준이라는 것.)

)

button.pack()
window.mainloop()
```

```python
#엔트리 위젯이 있는 윈도우를 생성해보자.
#딱 한 줄만 입력받는 역할인 위젯임.

from tkinter import *
window = Tk()

entry = Entry(window,
              
        fg = "#000000", #입력받을 글자의 색
        bg = "#F5FF2D", #배경색
        width = 20 #크기 (글자수 기준이라는 것.)

)

entry.pack()
window.mainloop()
```

```python
#압축 배치 관리자

from tkinter import *
window = Tk()

label1 = Label(window, text = "박스 #1" ,bg = "red", fg = "white").pack(side = RIGHT)
label2 = Label(window, text = "박스 #2", bg = "#fbff00", fg = "#000000").pack(side = LEFT)
label3 = Label(window, text = "박스 #3", bg = "#000000", fg = "#FFFFFF").pack()

window.mainloop()
```
